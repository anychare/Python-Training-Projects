## SECTION A: Python Basics [10]

### 1. Write a program to display all odd numbers between a given range. [5]

```python
rs = int(input('Enter the start of the range:'))
re = int(input('Enter the end of the range:'))
print(f'\nOdd numbers between {rs} and {re}:')
for i in range(rs,re+1):
    if i%2==1:
        print(i, end = ' ')
    else:
        continue       
```

### 2. Write a program to find the total count of a substring in a given string ignoring the case. (e.g. find the total count of the substring "data" in the string "I am learning data analytics so that I can be a Data Analytics professional". [5]


```python
st = input('Enter a string')
sub_string = input('Enter the substring to count')

# convert both to lower case
temp_str = st.lower()
count = temp_str.count(sub_string.lower())

print("The count of the substring is:", count)
```

### 3. Write a program to store 10 random integers between 1 and 20 into a list. Display the list. [6]

```python
from random import randint
my_list = []
for i in range(1,10+1):
    my_list.append(randint(1,20))

print('A list of random numbers between 1 and 20:\n', my_list)
```

### 4. Find the highest and lowest integers in the list. [4]

```python
highest_number = max(my_list)
print('The highest number if the list is:',highest_number)
```

```python
lowest_number = min(my_list)
print('Lowest number in the list is:', lowest_number)
```

### 5. Load the 'Interns_Selection' dataset (sent on the group) into a dataframe called 'interns_df'. Wrangle and clean the dataset as follows: 
#### a. Remove duplicates (if any). [4]
#### b. If there are any empty numeric cells, fill them in with the value 10. [4]
#### c. Use the one-hot encoding technique to encode the non-numeric columns into numeric columns as necessary. [5]
#### d. Export the clean dataset to Excel with the name 'Internship_Selection'. [2]

```python
import pandas as pd
interns_df = pd.read_csv(r"C:\$_ORT DATA\Training\Python\Sets\Interns_Selection.csv")
```

```python
#Exploring the dataframe information
interns_df.info()
```

```python
#Check for duplicates
interns_df.duplicated().any()
```

```python
#Display the duplicated rows
interns_df[interns_df.duplicated()]
```

```python
#Display the duplicated rows including the first copy
interns_df[interns_df.duplicated(keep=False)]
```

```python
#a. Remove duplicates
interns_df.drop_duplicates(inplace = True)
```

```python
#b. Replacing all empty numeric cells with 10
interns_df = interns_df.fillna(10)
```

```python
#c. Encoding non-numeric cells using one-hot encoding
en_interns_df = pd.get_dummies(interns_df, columns=['extracurricular', 'college_tier', 'placement_training'], dtype=int)
en_interns_df.info()
```

```python
#Exporting the clean dataframe to Excel
en_interns_df.to_excel('Internship_Selection.xlsx')
```

### 6. The number of students in the dataset. [2]

```python
import pandas as pd
df = pd.read_excel('Internship_Selection.xlsx')
number_of_students = df['student_id'].count()
print('Number of students:', number_of_students)
```

### 7. How many students were selected? [2]

```python
not_selected = df[(df['selected']==0)].reset_index()
print('Number of students not selected:', len(not_selected))
```

### 9. How many selected students have more than 5 certifications? [4]

```python
print(len(df[(df['selected']==1) & (df['certifications_count']>5)]))
```

### 10. Visualise the distribution of the number of internships done by the students. Also display the plotted figures below the graph. [10] 

```python
import seaborn as sb
import matplotlib.pyplot as plt
plt.figure(figsize = (11,6))
intern_dist = df['internships_done'].value_counts()
intern_dist.plot(kind = 'bar')
plt.title('Internship Distribution')
plt.xlabel('Internships Done')
plt.ylabel('Number of Students')
plt.show()
print(intern_dist)
```

### 11. Visualise the students who were selected vs those who were not selected. [5]

```python
plot = sb.countplot(data = df, x = 'selected')
plot.bar_label(plot.containers[0])
plt.show()
```

```python
plot = sb.countplot(data = df, x = 'selected', hue = 'selected')
for container in plot.containers:
    plot.bar_label(container)
plt.show()
```
