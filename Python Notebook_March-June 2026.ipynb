### 2. Write a program to calculate the factorial of a given number. Use a function. The factorial of a number n = nx(n-1)x(n-2)x(n-3)x.... e.g Factorial of 4 = 4x3x2x1

```python
#Defining the function
def factorial(n):
    if n<0:
        return "Cannot find factorial of a negative number"
    else: 
        fact = 1
        for i in range(1,n+1):
            fact*=i
        return fact

#Accepting input
number = int(input('Enter the number;'))
print(f"Factorial of {number} is", factorial(number))
```

## Checking for Outliers in a Dataframe 

### Using Interquartile Range (IQR)

```python
#Import pandas
import pandas as pd
#Creating sample data
df = pd.DataFrame({'Age': [20,25,22,100,25,20,4]})
#Work out the quartiles (Q1, Q2, Q3)
Q1 = df['Age'].quantile(0.25)
Q2 = df['Age'].quantile(0.50)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1
#Define the lower boundary and the upper boundary
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
#Identify the outliers
outliers = df[(df['Age'] < lower_bound) | (df['Age']> upper_bound)]
outliers
```

### Using the Z-score Method

```python
#Import libraries 
import pandas as pd
import numpy as np
from scipy import stats
#Creating sample data
df = pd.DataFrame({'Age': [20,25,22,100,25,20]})
#Calculate z-scores 
z_scores = np.abs(stats.zscore(df['Age']))
#Extract outliers where Z-score > 3
z_outliers = df[z_scores > 3]
z_outliers
```

## K-Nearest Neighbor Algorithm

### Example: Fraud classification model using KNN

```python
#Import Pandas and load dataset
import pandas as pd
df = pd.read_excel('Random.xlsx')
df.head(2)
```

```python
#Profile thedataset
df.info()
```

```python
#Select features
x = df.drop(columns = 'Fraud Class')
y = df[['Fraud Class']]
```

```python
#Split data into training and testing sets
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42)
```

```python
#Standardise/ normalise data
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
#Create the KNN model
from sklearn.neighbors import KNeighborsClassifier
k = 8
knn = KNeighborsClassifier(n_neighbors = k)
knn
```

```python
#Train the model
knn.fit(x_train, y_train)
knn
```

```python
#Test the model using the test set
predicted_fraud_class = knn.predict(x_test)
```

```python
#Evaluate the model accuaracy
from sklearn.metrics import accuracy_score
acc_score = accuracy_score(y_test, predicted_fraud_class)
print(acc_score)
```

```python
#Cross validation to choose the best k value
from sklearn.model_selection import cross_val_score
import numpy as np
k_values = [i for i in range(1,31)]
scores = []
scaler = StandardScaler()
X = scaler.fit_transform(x)

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors = k)
    score = cross_val_score(knn, X, y, cv = 5)
    scores.append(np.mean(score))
```

```python
#Plot the scores on a line graph
from matplotlib import pyplot as plt
import seaborn as sb
sb.lineplot(x = k_values, y = scores, marker = 'o', markerfacecolor = 'm')
plt.title('K Values Plot')
plt.xlabel('K-Values')
plt.ylabel('Accuracy Scores')
plt.show()
```

```python
#Displaying the value of the best k 
best_k_value = np.argmax(scores)
print(best_k_value)
```

```python
#Retrain the model using the best k value
knn = KNeighborsClassifier(n_neighbors = best_k_value)
knn.fit(x_train, y_train)
```

```python
#Re-test the model
new_predicted_fraud = knn.predict(x_test)
```

```python
#Re-evaluate the model accuracy using more metrics (accuracy score, precision score, recall score)
from sklearn.metrics import accuracy_score, recall_score, precision_score
p_score = precision_score(y_test, new_predicted_fraud)
a_score = accuracy_score(y_test, new_predicted_fraud)
r_score = recall_score(y_test, new_predicted_fraud)
print('Accuracy Score:', a_score)
print('Precision Score:', p_score)
print('Recall Score:', r_score)

```
