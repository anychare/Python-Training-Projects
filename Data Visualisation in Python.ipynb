```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_excel("C:\Data\ORT SA\Training\Power BI\Sets\Dataset_clean.xlsx")
data
```

### Plot a bar graph to visualise sales by region to indicate the region with the most sales.

```python
#Group all similar regions together
grouped_data = data.groupby('Region').sum().reset_index()

#Create x and y axes
x = grouped_data['Region'].to_list()
y = grouped_data['Sales'].to_list()

#Plot the bar graph using x and y above
plt.bar(x,y, color ='m', label = 'Sales')
plt.title('Sales by Region Report')
plt.xlabel('Region')
plt.ylabel('Sales')
for i in range(len(x)):
    plt.text(i, y[i], y[i], ha = 'center')
plt.savefig('Downloads/Barchart.png')
plt.show()
```

### Use a pie chart to show the contribution of each salesperson to the total sales. Who contributed the most according to the pie chart?

```python
grouped_df = data.groupby('Last Name').sum().reset_index()
labeldata = grouped_df['Last Name'].to_list()
values = grouped_df['Sales'].to_list()
plt.pie(values, labels = labeldata, autopct = '%.3f%%')
plt.show()
```

### Connect to the 'Diabetes' dataset and answer the following question: 
#### 1. What is the distribution of patients' age? Use a graph to show the distribution.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv(r"C:\Data\ORT SA\Training\Python\Sets\Diabetes.csv")
df.head()
```

```python
age = df['age'].tolist()
plt.figure(figsize = (11,6))
bins = 30
plt.hist(age, bins, histtype = 'bar', color = 'm')
plt.title('Patient Age Distribution')
plt.xlabel('Age')
plt.ylabel('Age Frequency')
plt.show()
```

```python
plt.figure(figsize = (11,6))
age_dist = df['age'].value_counts()
age_dist.plot(kind = 'bar')
plt.title('Patient Age Distribution')
plt.xlabel('Age')
plt.ylabel('Age Frequency')
plt.show()
print(age_dist)
```

### What are the discharge disposition IDs of the patients readmitted within 30 days? Show in a graph.

```python
discharge_disp = df[df['readmitted'] == '<30'] ['discharge_disposition_id'].value_counts()
discharge_disp
```

```python
plt.figure(figsize = (11,6))
discharge_disp.plot(kind = 'bar')
plt.title('Discharge Dispositions Report')
plt.xlabel('Discharge Disposition')
plt.ylabel('Number of Patients')
plt.show()
```
